## 2. Causal Masking for Auto-Regressive Generation
When generating text, a token cannot look ahead at future tokens. We enforce this by applying a causal mask (an upper triangular matrix) to the attention scores before the softmax operation, replacing future connections with $-\infty$.

```
seq_len = 4
scores = np.random.randn(seq_len, seq_len)

# Create an upper triangular matrix (1s above the diagonal)
mask = np.triu(np.ones((seq_len, seq_len)), k=1)
print("Causal Mask (1 indicates a masked position):\n", mask)

# TODO 2: Use boolean indexing to set the elements of 'scores' to -np.inf where mask == 1.
# SOLUTION: scores[mask == 1] = -np.inf
# YOUR CODE HERE

print("\nMasked Scores:\n", scores)
```

## 3. Visualizing Multi-Head Attention Patterns
Let's visualize how different attention heads might focus on different parts of a sentence. For example, Head 1 might look at adjacent words, while Head 2 might look at the end of the sentence.

```
import matplotlib.pyplot as plt
import seaborn as sns

# Dummy attention weights for two different heads (4x4 sequence)
head1_weights = np.eye(4) # Focuses on itself
head2_weights = np.triu(np.ones((4,4))) / 4 # Focuses on future (bad pattern, but illustrative)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

sns.heatmap(head1_weights, ax=axes[0], cmap='Blues', annot=True, cbar=False)
axes[0].set_title('Head 1: Diagonal Focus')

sns.heatmap(head2_weights, ax=axes[1], cmap='Oranges', annot=True, cbar=False)
axes[1].set_title('Head 2: Broad Focus')

plt.tight_layout()
plt.show()
```
